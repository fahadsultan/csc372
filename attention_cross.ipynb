# Cross-Attention

<img src="assets/Chap12/TransformerBlockSACross.svg" style="filter: invert(1);" width="100%">
Cross-attention. The flow of computation is the same as in standard
self-attention, but the queries are calculated from the decoder embeddings Xdec,
and the keys and values from the encoder embeddings Xenc. For translation
tasks, the encoder contains information about the source language statistics, and
the decoder contains information about the target language statistics.

---

**Cross-attention** connects the target-language translation being generated with the source-language sentence. The computation follows the same steps as self-attention, but the inputs come from two different places:

- **Queries** come from representations of the **target-language words generated so far**.
- **Keys and values** come from representations of the **source-language sentence**.

Think of a query as a question: **“What information from the source sentence would help me predict the next target-language word?”** The model compares each query with the source-language keys. These comparisons produce attention weights, which indicate how much attention to give each source position. The model then uses those weights to combine the corresponding values.

For example, suppose we are translating the English sentence “The cat is sleeping” into French. When predicting the French word “chat,” the model can give more attention to the representation of “cat” in the English sentence. When predicting another word, it can focus on a different part of the sentence. Attention can also be spread across several words when their combined meaning is useful.

The key difference is **where the model looks**. Self-attention over the target language looks at the translation generated so far. Cross-attention looks at the source-language sentence. Together, these help the model produce a translation that makes sense in the target language while preserving the meaning of the source.

During training, the model learns patterns in both languages. During translation, cross-attention connects information about the **actual source sentence** with the **target-language translation generated so far**.
