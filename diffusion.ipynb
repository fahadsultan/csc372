## Overview 

<img src="assets/Chap18/DiffusionOverview.svg" width="100%" style="filter:invert(1)">
Diffusion models. The encoder (forward, or diffusion process) maps
the input x through a series of latent variables z1 . . . zT . This process is prespecified
and gradually mixes the data with noise until only noise remains. The
decoder (reverse process) is learned and passes the data back through the latent
variables, removing noise at each stage. After training, new examples are
generated by sampling noise vectors zT and passing them through the decoder.
