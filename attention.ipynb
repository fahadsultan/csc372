# Attention 




To motivate the need for attention mechanisms, consider the following passage:

> The <mark>restaurant</mark> refused to serve me a ham sandwich because <mark>it</mark> only serves vegetarian
food. In the end, they just gave me two slices of bread. <mark>Their</mark> ambiance was just as good
as the food and service.

The goal is to design a network to process this text into a representation suitable for downstream tasks. 

For example, it might be used to classify the review as positive or negative or to answer questions such as “Does the restaurant serve steak?”.

Three properties of language motivate attention:

1. <mark>Inputs vary in length.</mark> **A model should process both a short sentence and a long passage using the same learned parameters.**

2. <mark>Meaning depends on context.</mark> A token embedding initially represents a token without incorporating the surrounding passage. To interpret “it” in this example, the model needs information about “restaurant.”

3. <mark>Relevant information can be far away.</mark> “Their” in the final sentence refers back to “restaurant.” The model needs a way to combine information across positions, even when those positions are separated by many tokens.

A naive approach would flatten all token embeddings into one large vector and pass it through a fully connected network. This ties the architecture to a particular input length and can require many parameters. Attention instead applies shared transformations at every position and learns how to combine information from different positions.

Attention produces a new representation for each token by taking a weighted sum of information from other tokens. The weights depend on the input: a token can draw on different parts of the passage in different contexts.

Parameter sharing does not eliminate every computational cost. Standard dense self-attention computes a score for every pair of positions, so the number of scores grows quadratically with sequence length.

The following table illustrates various linguistic phenomena and how they connect to what a model needs to learn in order to handle them correctly.


| Linguistic phenomenon	| Example	| Connection to learn |
|----------------------|-----------------|-------------------|
| Pronoun resolution	| The dog chased the cat because it was frightened.	| What does it refer to? |
| Pronoun resolution	| Sarah called Emma after she finished work.	| Which person finished work? |
| Subject–verb agreement	| The keys to the old cabinet are missing.	| Are must agree with keys, not cabinet. |
| Adjective modification	| The old man carried a heavy suitcase.	| Which noun does each adjective describe? |
| Adverb modification	| She barely passed the exam.	| Barely changes how we interpret passed. |
| Negation	| The movie was not particularly interesting.	| Not reverses the evaluation. |
| Word-sense disambiguation	| He sat by the bank and watched the river.	| River helps determine the meaning of bank. |
| Word-sense disambiguation	| She deposited money at the bank.	| Money and deposited indicate a financial institution. |
| Long-distance dependency	| The book that the professor recommended yesterday was fascinating.	| Was relates to book, despite intervening words. |
| Semantic roles	| The chef cooked the fish with a pan.	| Pan is an instrument used for cooking. |
| Semantic roles	| The chef cooked the fish with lemon.	| Lemon is an ingredient or accompaniment. |
| Contrast	| Although the restaurant was expensive, the food was excellent.	| Although signals a contrast between clauses. |
| Coreference across sentences	| John bought a bicycle. He rode it home.	| He connects to John; it to bicycle. |
| Conditional relationships	| If it rains, we will cancel the picnic.	| If establishes a dependency between clauses. |

The central teaching point is that a token's representation should depend not only on the token itself, but also on other relevant tokens in the sequence.
